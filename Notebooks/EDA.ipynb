VISUALIZATION

```python
plt.figure(figsize=(5,4))
sns.countplot(x='heart_disease_present',data=df)
plt.title("Heart Disease Distribution")
plt.xlabel("Heart Disease Present")
plt.ylabel("Count")
plt.show()
```

Correlation

```python
df_numeric = df.drop(columns=['patient_id'])
```

```python
df_numeric = pd.get_dummies(df_numeric, drop_first=True)
```

```python
plt.figure(figsize=(12,10))
sns.heatmap(df_numeric.corr(), cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

A correlation heatmap was generated to examine the relationships among the features. Most feature pairs show weak to moderate correlations, with no evidence of severe multicollinearity. The diagonal values are 1 because each feature is perfectly correlated with itself

Histograms

```python
sns.boxplot(x='heart_disease_present', y='age', data=df)
plt.show()
```

Patients with heart disease tend to be older than those without heart disease. The median age is higher for the heart disease group, indicating that age may be an important factor in predicting heart disease. However, since the age distributions overlap, age should be considered along with other features.
